```python
dbutils.widgets.text("table_name", "")
dbutils.widgets.text("source_system", "")
dbutils.widgets.text("landing_timestamp", "")
table_name = dbutils.widgets.get("table_name")
source_system = dbutils.widgets.get("source_system")
landing_timestamp = dbutils.widgets.get("landing_timestamp")
```

```python
landing_path=f"/Volumes/caresync_test2/landing/landing_volume/{source_system}/{table_name}/{landing_timestamp}/"
```

```python
print(landing_path)
```

```python
dbutils.fs.ls('/Volumes/caresync_test2/landing/landing_volume/neon_postgres/hospitals/2026-09-03T11:22:49.5508469Z')
```

```python
landing_df=spark.read.format('parquet').load(landing_path)
```

```python
%skip
display(landing_df)
```

```python
from pyspark.sql.functions import lit, current_timestamp

udpated_df=landing_df.withColumn("landing_timestamp",lit(landing_timestamp)).withColumn("insert_timestamp",current_timestamp()).withColumn("update_timestamp",current_timestamp())

```

```python
udpated_df.write.mode("overwrite").saveAsTable(f"caresync_test2.bronze.{table_name}")
```
